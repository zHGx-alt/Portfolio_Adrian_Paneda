```python
import pandas as pd
```

```python
df = pd.read_csv("/home/zhgx/Portfolio_Adrian_Paneda/ecommerce-analytics-project/data_sample/events_tiny.csv")
df_clean = df.copy()
```

```python
# 1. Parse datetime
df_clean["event_time"] = pd.to_datetime(df_clean["event_time"], errors="coerce")

# 2. Drop rows with invalid datetime
df_clean = df_clean.dropna(subset=["event_time"])

# 3. Fix user_id
df_clean = df_clean.dropna(subset=["user_id"])
df_clean["user_id"] = df_clean["user_id"].astype("int64")

# 4. Keep only useful events
df_clean = df_clean[df_clean["event_type"].isin(["view", "cart", "purchase"])]

# 5. Fill missing category_code and brand
df_clean["category_code"] = df_clean["category_code"].fillna("unknown")
df_clean["brand"] = df_clean["brand"].fillna("unknown")

# 6. Remove invalid prices
df_clean = df_clean[df_clean["price"] > 0]

# Save cleaned data
df_clean.to_csv(
    "/home/zhgx/Portfolio_Adrian_Paneda/ecommerce-analytics-project/data_sample/clean_data.csv",
    index=False
)
```
