# Project Report: Machine Learning Applied to Cryptocurrencies

## 1. Introduction

The objective of this project is to apply **Machine Learning** techniques to historical cryptocurrency data in order to analyze their behavior and evaluate the ability of different models to perform predictions.

The project was developed in a **Jupyter Notebook**, integrating exploratory analysis, data preprocessing, and model training, with the aim of following a clear and reproducible workflow.

---

## 2. Data Loading and Understanding

First, the data corresponding to different cryptocurrencies were loaded.  
During this initial phase, the following aspects were analyzed:

- The structure of the dataset (rows and columns).  
- The meaning of each variable.  
- Data types.  
- The presence of missing or inconsistent values.

This initial analysis made it possible to understand the temporal nature of the data and their suitability for prediction tasks.

---

## 3. Data Cleaning and Preprocessing

Subsequently, the data cleaning and preparation process was carried out, which included:

- Handling of missing values.  
- Conversion of temporal columns to appropriate formats.  
- Removal of low-relevance or redundant variables.  
- Creation of new useful features when necessary.

This step was essential to ensure that the data used by the models were consistent and of high quality.

---

## 4. Exploratory Data Analysis (EDA)

Once the data were prepared, an **Exploratory Data Analysis (EDA)** was performed to:

- Study the price evolution of cryptocurrencies.  
- Analyze the distribution of the variables.  
- Identify possible correlations between variables.  
- Detect relevant patterns or behaviors in the market.

Visualizations and descriptive statistics helped to better understand the volatility and dynamics inherent to cryptocurrencies.

---

## 5. Preparation for Machine Learning

Before training the models, the following steps were performed:

- Selection of predictor variables and the target variable.  
- Split of the dataset into training and test sets.  
- Scaling or normalization of variables when required.  
- Adaptation of the data to a suitable format for Machine Learning algorithms.

This process made it possible to build a coherent pipeline for model training.

---

## 6. Model Training and Evaluation

Next, one or more **Machine Learning** models were trained on the training data.  
To evaluate their performance, appropriate metrics were used for the type of problem, comparing the predicted values with the true values.

This analysis made it possible to assess the ability of the models to capture the behavior of the cryptocurrency market.

---

## 7. Results and Analysis

The obtained results show the performance of the applied models and allow conclusions to be drawn regarding:

- The difficulty of predicting prices in highly volatile markets.  
- The influence of the selected variables.  
- The limitations of the models used.

Both correct predictions and model errors are analyzed.

---

## 8. Conclusions

In conclusion, this project demonstrates the practical application of **Machine Learning** techniques in the cryptocurrency domain, highlighting:

- The importance of proper data preprocessing.  
- The value of prior exploratory analysis.  
- The inherent difficulties of prediction in financial markets.

As possible future improvements, more advanced models could be tested, additional historical data could be incorporated, or external variables influencing the market could be included.
